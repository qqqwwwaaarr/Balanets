```python
from ColabTurtle.Turtle import *
initializeTurtle()
speed(13)

def draw_heart():
    color("red")
    pensize(3)

    left(140)
    forward(180)

    # levá křivka
    for _ in range(200):
        right(1)
        forward(1)

    left(120)

    # pravá křivka
    for _ in range(200):
        right(1)
        forward(1)

    forward(180)

clear()
draw_heart()
```

```python
from ColabTurtle.Turtle import *
initializeTurtle()
speed(13)

color('blue')
width(3)

# Draw a dashed circle
segments = 72 # Number of segments for the circle
for _ in range(segments):
    pendown()
    forward(5) # Dash length
    penup()
    forward(5) # Gap length
    left(360 / segments) # Turn to form a circle
```

```python
from ColabTurtle.Turtle import *
import random

initializeTurtle()
speed(10)
bgcolor('black')

colors = ["red", "orange", "yellow", "lime", "cyan", "blue", "magenta", "white"]

window_w, window_h = window_size

def draw_circle(radius):
    for _ in range(36):
        forward( (2 * 3.14 * radius) / 36 )
        left(10)

clear()

for _ in range(30):       # počet kroužků
    # náhodná pozice
    x = random.randint(50, window_w - 50)
    y = random.randint(50, window_h - 50)

    penup()
    goto(x, y)
    pendown()

    # náhodná barva
    pencolor(random.choice(colors))

    # náhodný poloměr
    r = random.randint(10, 70)

    draw_circle(r)
```

```python
from ColabTurtle.Turtle import *

initializeTurtle(initial_speed=13)
bgcolor("black")
speed(13)

# Rainbow barvy
colors = [
    "#FF0000",
    "#FFA6D3",
    "#FFFF00",
    "#00FF00",
    "#0000FF",
    "#4B0082",
    "#8B00FF"
]

for i in range(1, 250):   # začínáme od 1
    color(colors[i % len(colors)])
    width((i % 5) + 1)

    forward(i * 0.8)
    left(35)

    # simulace malého kruhového oblouku místo circle()
    for _ in range(6):
        forward(0.2 * i)
        left(6)

hideturtle()



```

```python


from ColabTurtle.Turtle import *

initializeTurtle(initial_speed=13)
bgcolor("black")
speed(13)

# Rainbow barvy
colors = [
    "#FF66FF",
    "#FFFFFF",
    "#FFFF00",
    "#33FF99",
    "#2B00FF",
    "#4B0082",
    "#FF0080"
]

num_spirals = 12
angle_between = 360 / num_spirals

for n in range(num_spirals):
    penup()
    home()
    setheading(n * angle_between)
    pendown()

    step = 0.6
    turn_angle = 46


    for i in range(1, 150):  # začínáme od 1
        pencolor(colors[i % len(colors)])
        width((i % 5) + 1)

        forward(i * step)
        left(turn_angle)
        # místo circle() simulujeme oblouk malými krokami
        for _ in range(6):
            forward(0.2 * i)
            left(6)

hideturtle()
```

```python
!pip install ColabTurtle

from ColabTurtle.Turtle import *

initializeTurtle(initial_speed=13)
bgcolor("lightblue")
speed(10)

# --- Rybička ---
penup()
goto(-50, 0)
pendown()
pencolor("orange")

# Tělo rybičky (elipsa)
for _ in range(2):
    forward(100)
    left(90)
    forward(50)
    left(90)

# Ocas rybičky (trojúhelník)
penup()
goto(-150, 0)
pendown()
goto(-180, 30)
goto(-180, -30)
goto(-150, 0)

# --- Bublinky ---
pencolor("blue")
penup()
bubble_positions = [(-20, 30), (0, 50), (30, 70), (60, 60), (80, 40)]
for pos in bubble_positions:
    goto(pos)
    pendown()
    for _ in range(36):  # nakreslíme kruh simulací
        forward(5)
        left(10)
    penup()

hideturtle()
```

```python
import matplotlib.pyplot as plt

# Data cen nemovitostí v mil. Kč
prices_mil_kc = [2.1, 2.5, 3.0, 3.2, 3.8, 4.0, 4.5, 5.0]

# Vytvoření histogramu
plt.hist(prices_mil_kc, bins=5, edgecolor='black') # Můžete upravit počet 'bins' (přihrádek)

# Přidání titulu a popisků os
plt.title('Histogram cen nemovitostí')
plt.xlabel('Cena (mil. Kč)')
plt.ylabel('Frekvence')

# Zobrazení histogramu
plt.show()
```

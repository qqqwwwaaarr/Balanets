```python
class Kyticka:
    def __init__(self, druh, vyska, ma_kvet, cena):
        self.druh = druh
        self.vyska = vyska  # v cm
        self.ma_kvet = ma_kvet  # True/False
        self.cena = cena  # v Kč

    def info(self):
        stav = "má květ" if self.ma_kvet else "zatím nekvete"
        print(f"Kytička {self.druh} je vysoká {self.vyska} cm, {stav} a stojí {self.cena} Kč.")

    def zalij(self, kolik_cm):
        self.vyska += kolik_cm
        print(f"Kytička {self.druh} byla zalita a povyrostla o {kolik_cm} cm.")

    def rozkveti(self):
        if not self.ma_kvet:
            self.ma_kvet = True
            print(f"Kytička {self.druh} právě rozkvetla! 🌸")
        else:
            print(f"Kytička {self.druh} už dávno kvete.")

    def zmen_cenu(self, nova_cena):
        self.cena = nova_cena
        print(f"Cena kytičky {self.druh} byla aktualizována na {self.cena} Kč.")

# Vytvoření instance
moje_kytka = Kyticka("Tulipán", 15, False, 50)

# Výpis počátečního stavu
moje_kytka.info()

# Péče a úprava ceny
moje_kytka.zalij(5)
moje_kytka.rozkveti()
moje_kytka.zmen_cenu(65)

# Výpis konečného stavu
moje_kytka.info()
```
